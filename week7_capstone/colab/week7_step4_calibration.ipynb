# Week 7 capstone: Step 4, step-budget calibration (resume: pre-declared expansion to S=8000)

**Where calibration stands.** Stage 1 ran in the first calibration session (arm A, seed 900, 1000/2000/4000 steps, results are in the repo). The training-loss rule found **no grid value that qualifies** (final training loss was still improving by 9.2% and then 6.4% per doubling, against the 2% threshold). The pre-registration says what happens next: add S=8000 for arm A, only if the timing projection allows. It does (Section 23 of `EXPERIMENT_PLAN.md`).

**This notebook runs that expansion and whatever the rule then needs:** arm A at 8000 steps (one run, about 5 minutes). If 4000 to 8000 improves final training loss by less than 2%, S*=4000 and arms B and C are checked at 4000 and 8000 (four more runs). If it does not, the rule stops and the fallback (use the largest value, record non-convergence) is a human decision. The driver resumes: the three finished runs are skipped.

**Unchanged:** training loss only; arms B and C compute no validation or test metric; policy P1; no `pip install`; no pilot or main cells; gates checked before and during every run.
